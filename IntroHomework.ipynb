## 6. Median value of horsepower (1 point)

```python
# Step 1: Find the median value of horsepower column
median_before = df["horsepower"].median()
print("Median horsepower (before filling):", median_before)

# Step 2: Calculate the most frequent value of horsepower column
most_frequent_horsepower = df["horsepower"].mode()[0]
print("Most frequent horsepower value:", most_frequent_horsepower)

# Step 3: Fill missing values in horsepower with the most frequent value
df["horsepower"] = df["horsepower"].fillna(most_frequent_horsepower)

# Step 4: Calculate the median value of horsepower again
median_after = df["horsepower"].median()
print("Median horsepower (after filling):", median_after)

# Step 5: Compare the values
print("Median changed:", median_before != median_after)
if median_before != median_after:
    if median_after > median_before:
        print("Answer: Yes, it increased")
    else:
        print("Answer: Yes, it decreased")
else:
    print("Answer: No")
```

## 7. Sum of weights (1 point)

```python
# Step 1: Select all cars from Asia
asia_cars = df[df['origin'] == 'Asia']

# Step 2: Select only columns vehicle_weight and model_year
asia_selected = asia_cars[['vehicle_weight', 'model_year']]

# Step 3: Select the first 7 values
asia_first_7 = asia_selected.head(7)
print("First 7 cars from Asia with vehicle_weight and model_year:")
print(asia_first_7)

# Step 4: Get the underlying NumPy array
X = asia_first_7.values
print("NumPy array X:")
print(X)

# Step 5: Compute matrix-matrix multiplication: X.T @ X
XTX = X.T @ X
print("X.T @ X (XTX):")
print(XTX)

# Step 6: Invert XTX
XTX_inv = np.linalg.inv(XTX)
print("Inverse of XTX:")
print(XTX_inv)

# Step 7: Create array y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
print("Array y:")
print(y)

# Step 8: Compute w = (X.T @ X)^-1 @ X.T @ y
w = XTX_inv @ X.T @ y
print("Result w:")
print(w)

# Step 9: Get the sum of all elements of w
sum_w = np.sum(w)
print("Sum of all elements in w:", sum_w)
```
